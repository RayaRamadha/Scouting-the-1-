```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

df = pd.read_csv('FC26_20250921.csv', low_memory=False)
outfield = df[~df['player_positions'].str.contains('GK', na=False)].copy()
```

```python
all_vars = df.columns.tolist()
print(f"Total variabel: {len(all_vars)}")
print(all_vars)
```

```python
scouting_pool = outfield[
    (outfield['age'] <= 24) &
    (outfield['value_eur'] < 15000000)
].copy()
```

```python
scaler = MinMaxScaler()
metrics = ['movement_sprint_speed', 'power_strength', 'power_jumping']
scouting_pool[['spd_n', 'str_n', 'jmp_n']] = scaler.fit_transform(scouting_pool[metrics])
```

```python
scouting_pool['dominance_index'] = scouting_pool[['spd_n', 'str_n', 'jmp_n']].mean(axis=1)
top_specimens = scouting_pool.sort_values('dominance_index', ascending=False).head(5)
```

```python
print("="*60)
print("OUTLIER REPORT: THE PHYSICAL 1% (FC 26 PLAYSTYLES)")
print("="*60)

for index, row in top_specimens.iterrows():

    active_ps = [col.replace('player_playstyle_', '').replace('_', ' ').title()
                 for col in df.columns if 'player_playstyle_' in col and row[col] == 1]

    playstyle_display = ", ".join(active_ps) if active_ps else "Physical Specialist (Generic)"

    print(f"IDENTIFIED SUBJECT: {row['long_name'].upper()}")
    print(f"CLUB: {row['club_name']} | POS: {row['player_positions']}")
    print(f"MARKET VALUE: €{row['value_eur'] / 1000000:.2f}M")
    print(f"PLAYSTYLES: {playstyle_display}")
    print("-" * 30)
    print(f"SPD: {row['movement_sprint_speed']:.0f} | STR: {row['power_strength']:.0f} | JMP: {row['power_jumping']:.0f}")
    print(f"DOMINANCE INDEX: {row['dominance_index']:.4f}")
    print("="*60)
```
